```python
import math as mt

n = 10
r = 4

# Fórmula de permutaciones: P(n, r) = n! / (n - r)!
Pr = mt.factorial(n) / mt.factorial(n - r)

print(f"Total de asistentes (n): {n}")
print(f"Preguntas a calificar (r): {r}")
print(f"\nFórmula: P(n, r) = n! / (n - r)!")
print(f"P({n}, {r}) = {n}! / ({n} - {r})!")
print(f"P({n}, {r}) = {mt.factorial(n)} / {mt.factorial(n - r)}")
print(f"\nEl número de permutaciones es: {int(Pr)}")
print(f"\nExisten {int(Pr)} maneras de elegir los asistentes para calificar.")
```
