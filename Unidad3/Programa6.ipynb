```python
import numpy as np
import matplotlib.pyplot as plt

# 1. Definición de los datos
x = np.array([0, 1, 2, 3, 4, 5])
p_x = np.array([0.10, 0.25, 0.30, 0.20, 0.10, 0.05])
suma_prob = np.sum(p_x)
print(f"1. Verificación: La suma de las probabilidades es: {suma_prob:.2f}")

esperanza = np.sum(x * p_x)

varianza = np.sum((x**2) * p_x) - (esperanza**2)

desviacion = np.sqrt(varianza)

print(f"\n3. Cálculos:")
print(f"   - Valor Esperado (E[X]): {esperanza:.2f} células")
print(f"   - Varianza: {varianza:.4f}")
print(f"   - Desviación Estándar: {desviacion:.4f}")

f_x = np.cumsum(p_x)

# Graficar
plt.figure(figsize=(12, 5))

# Gráfico de la Distribución de Probabilidad
plt.subplot(1, 2, 1)
plt.bar(x, p_x, color='skyblue', edgecolor='black', alpha=0.7)
plt.title('Distribución de Probabilidad P(X)')
plt.xlabel('Número de células (X)')
plt.ylabel('Probabilidad')
plt.xticks(x)
plt.grid(axis='y', linestyle='--', alpha=0.6)

# Gráfico de la Función Acumulada (Escalonada)
plt.subplot(1, 2, 2)
plt.step(x, f_x, where='post', color='salmon', lw=2, marker='o')
plt.title('Función de Distribución Acumulada F(X)')
plt.xlabel('Número de células (x)')
plt.ylabel('F(x)')
plt.xticks(x)
plt.ylim(0, 1.1)
plt.grid(axis='both', linestyle='--', alpha=0.6)

plt.tight_layout()
plt.show()
```
