```python
import numpy as np
import matplotlib.pyplot as plt

unidades = np.array([0, 1, 2, 3, 4, 5])
probabilidades = np.array([0.05, 0.10, 0.25, 0.30, 0.20, 0.10])

suma_p = np.sum(probabilidades)
print(f"1. Verificación: La suma de las probabilidades es: {suma_p:.2f}")
valor_esperado = np.sum(unidades * probabilidades)
varianza = np.sum((unidades**2) * probabilidades) - (valor_esperado**2)

desviacion = np.sqrt(varianza)

print(f"\n3. Cálculos de Ventas:")
print(f"   - Unidades promedio esperadas (E[X]): {valor_esperado:.2f} unidades")
print(f"   - Varianza: {varianza:.4f}")
print(f"   - Desviación Estándar: {desviacion:.4f} unidades")

acumulada = np.cumsum(probabilidades)

plt.figure(figsize=(12, 5))

# Gráfico de barras de la distribución
plt.subplot(1, 2, 1)
plt.bar(unidades, probabilidades, color='plum', edgecolor='indigo', alpha=0.8)
plt.title('Distribución de Probabilidad de Ventas Semanales')
plt.xlabel('Unidades Vendidas (X)')
plt.ylabel('P(X)')
plt.xticks(unidades)
plt.grid(axis='y', linestyle='--', alpha=0.5)

# Gráfico de la función acumulada
plt.subplot(1, 2, 2)
plt.step(unidades, acumulada, where='post', color='darkmagenta', marker='D', lw=2)
plt.title('Función de Distribución Acumulada F(X)')
plt.xlabel('Unidades Vendidas (x)')
plt.ylabel('F(x)')
plt.xticks(unidades)
plt.ylim(0, 1.1)
plt.grid(axis='both', linestyle=':', alpha=0.7)

plt.tight_layout()
plt.show()
```
