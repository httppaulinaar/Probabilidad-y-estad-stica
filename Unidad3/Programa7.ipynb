```python
import numpy as np
import matplotlib.pyplot as plt

x_premios = np.array([0, 10, 20, 50])
p_x_premios = np.array([0.50, 0.30, 0.15, 0.05])

suma_p = np.sum(p_x_premios)
print(f"1. Verificación: La suma de las probabilidades es: {suma_p:.2f}")
# Valor Esperado (E[X])
esperanza_premio = np.sum(x_premios * p_x_premios)

# Varianza y Desviación Estándar
varianza_premio = np.sum((x_premios**2) * p_x_premios) - (esperanza_premio**2)
desviacion_premio = np.sqrt(varianza_premio)

print(f"\n3. Cálculos de la Tómbola:")
print(f"   - Ganancia Promedio Esperada (E[X]): ${esperanza_premio:.2f} pesos")
print(f"   - Varianza: {varianza_premio:.2f}")
print(f"   - Desviación Estándar: ${desviacion_premio:.2f} pesos")
f_x_premios = np.cumsum(p_x_premios)

# Visualización
plt.figure(figsize=(12, 5))

# Distribución de Probabilidad
plt.subplot(1, 2, 1)
plt.bar(x_premios.astype(str), p_x_premios, color='mediumseagreen', edgecolor='black')
plt.title('Distribución de Probabilidad de Premios')
plt.xlabel('Monto del Premio ($)')
plt.ylabel('Probabilidad P(X)')
plt.grid(axis='y', alpha=0.3)

# Función Acumulada
plt.subplot(1, 2, 2)
plt.step(range(len(x_premios)), f_x_premios, where='post', color='darkorange', marker='s')
plt.xticks(range(len(x_premios)), x_premios)
plt.title('Función de Distribución Acumulada F(X)')
plt.xlabel('Monto del Premio ($)')
plt.ylabel('F(X)')
plt.ylim(0, 1.1)
plt.grid(alpha=0.3)

plt.tight_layout()
plt.show()
```
