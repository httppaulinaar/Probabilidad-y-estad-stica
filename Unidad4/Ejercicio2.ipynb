```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import norm
```

```python
# ============================================================
# EJERCICIO 2
# 500 estudiantes del ITD | Media = 70 kg | Desv. Std = 4 kg
# ============================================================

# Parámetros de la distribución normal
media = 70
desv_std = 4
total_estudiantes = 500

# ============================================================
# 1. CUÁNTOS ESTUDIANTES PESAN MENOS DE 75 KG
# ============================================================

peso_limite = 75

prob_menos_75kg = norm.cdf(peso_limite, media, desv_std)
estudiantes_menos_75kg = total_estudiantes * prob_menos_75kg

print("=" * 55)
print("1. ESTUDIANTES QUE PESAN MENOS DE 75 KG")
print("=" * 55)
print(f"Probabilidad de que un estudiante pese menos de 75 kg: {prob_menos_75kg:.4f}")
print(f"Número de estudiantes que pesan menos de 75 kg: {estudiantes_menos_75kg:.2f}")
```

```python
# Gráfica
plt.figure(figsize=(6, 4))
x = np.linspace(media - 4 * desv_std, media + 4 * desv_std, total_estudiantes)
y = norm.pdf(x, media, desv_std)
plt.plot(x, y, 'b', linewidth=2)
plt.fill_between(x, y, where=(x <= peso_limite), color='skyblue', alpha=0.5)
plt.title(f'Probabilidad de que un estudiante pese menos de 62 kg.\n $\\mu=70$, $\\sigma=4$, X=500')
plt.xlabel('Peso (kg)')
plt.ylabel('Densidad de probabilidad')
plt.grid(True)
plt.show()
```

```python
# ============================================================
# 3. CUÁNTOS ESTUDIANTES PESAN MÁS DE 79 KG
# ============================================================

peso_limite = 79

prob_menos_79kg = norm.cdf(peso_limite, media, desv_std)
prob_mas_79kg = 1 - prob_menos_79kg
estudiantes_mas_79kg = total_estudiantes * prob_mas_79kg

print("\n" + "=" * 55)
print("3. ESTUDIANTES QUE PESAN MÁS DE 79 KG")
print("=" * 55)
print(f"Probabilidad de que un estudiante pese más de 79 kg: {prob_mas_79kg:.4f}")
print(f"Número aproximado de estudiantes que pesan más de 79 kg: {estudiantes_mas_79kg:.2f}")
```

```python
# Gráfica
plt.figure(figsize=(6, 4))
x = np.linspace(media - 4 * desv_std, media + 4 * desv_std, total_estudiantes)
y = norm.pdf(x, media, desv_std)
plt.plot(x, y, 'b', linewidth=2)
plt.fill_between(x, y, where=(x >= peso_limite), color='skyblue', alpha=0.5)
plt.title(f'Probabilidad de que un estudiante pese mas de 79 Kg.\n $\\mu=70$, $\\sigma=4$, X=500')
plt.xlabel('Peso (kg)')
plt.ylabel('Densidad de probabilidad')
plt.grid(True)
plt.show()
```
