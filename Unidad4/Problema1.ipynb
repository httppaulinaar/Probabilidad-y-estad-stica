```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import norm
```

```python
# ============================================================
# EJERCICIO 3
# 500 estudiantes del ITD | Media = 70 kg | Desv. Std = 4 kg
# ============================================================

# Parámetros de la distribución normal
media = 70
desv_std = 4
total_estudiantes = 500

# ============================================================
# ESTUDIANTES QUE PESAN MÁS DE 68 Y MENOS DE 72 KG
# ============================================================

peso_limite_inferior = 68
peso_limite_superior = 72

# Calculamos la probabilidad acumulada hasta cada límite usando CDF
prob_menos_72kg = norm.cdf(peso_limite_superior, media, desv_std)
prob_menos_68kg = norm.cdf(peso_limite_inferior, media, desv_std)

# La probabilidad del intervalo es la diferencia entre ambas CDF
prob_entre_68_72kg = prob_menos_72kg - prob_menos_68kg

# Calculamos el número de estudiantes en ese rango
estudiantes_entre_68_72kg = total_estudiantes * prob_entre_68_72kg

print("=" * 60)
print("ESTUDIANTES QUE PESAN MÁS DE 68 Y MENOS DE 72 KG")
print("=" * 60)
print(f"P(X < 72 kg):               {prob_menos_72kg:.10f}")
print(f"P(X < 68 kg):               {prob_menos_68kg:.10f}")
print(f"P(68 < X < 72):             {prob_entre_68_72kg:.10f}")
print(f"Número de estudiantes:      {estudiantes_entre_68_72kg:.2f}")
```

```python
# ============================================================
# GRÁFICA
# ============================================================

plt.figure(figsize=(6, 4))

x = np.linspace(media - 4 * desv_std, media + 4 * desv_std, 1000)
y = norm.pdf(x, media, desv_std)

# Curva principal
plt.plot(x, y, 'b', linewidth=2)

# Área sombreada entre 68 y 72 kg
plt.fill_between(x, y,
                 where=((x >= peso_limite_inferior) & (x <= peso_limite_superior)),
                 color='skyblue', alpha=0.5,
                 label=f'P(68 < X < 72) = {prob_entre_68_72kg:.4f}')

# Líneas verticales en los límites
plt.axvline(x=peso_limite_inferior, color='red', linestyle='--', linewidth=1.5, label='68 kg')
plt.axvline(x=peso_limite_superior, color='green', linestyle='--', linewidth=1.5, label='72 kg')

plt.title(f'Probabilidad de que un estudiante pese entre 68 y 72 kg.\n'
          f'$\\mu=70$, $\\sigma=4$, X=500')
plt.xlabel('Peso (kg)')
plt.ylabel('Densidad de probabilidad')
plt.legend()
plt.grid(True)
plt.show()
```
